```julia
count_aa = 0  # Count strings containing "111"
count_ab = 0 # Count strings containing 1, 1, 1 anywhere
count_ba = 0 # Count strigns containing "123"
count_bb = 0  # Count strings containing 1, 2, and 3 in any order

query_a = 1
query_b = (1, 2, 3)

for (i,j,k,l,m,n) in Iterators.product(0:9, 0:9, 0:9, 0:9, 0:9, 0:9)
    # Check if contains three consecutive query_a's
    if (i == query_a && j == query_a && k == query_a) ||
       (j == query_a && k == query_a && l == query_a) ||
       (k == query_a && l == query_a && m == query_a) ||
       (l == query_a && m == query_a && n == query_a)
        count_aa += 1
    end

    # Inline counting instead of calling count() - use boolean addition
    count_a = (i == query_a) + (j == query_a) + (k == query_a) + (l == query_a) + (m == query_a) + (n == query_a)
    if count_a >= 3
        count_ab += 1
    end

    if (i == query_b[1] && j == query_b[2] && k == query_b[3]) ||
       (j == query_b[1] && k == query_b[2] && l == query_b[3]) ||
       (k == query_b[1] && l == query_b[2] && m == query_b[3]) ||
       (l == query_b[1] && m == query_b[2] && n == query_b[3])
        count_ba += 1
    end
    
    # Inline counting for all three digits
    count_b1 = (i == query_b[1]) + (j == query_b[1]) + (k == query_b[1]) + (l == query_b[1]) + (m == query_b[1]) + (n == query_b[1])
    count_b2 = (i == query_b[2]) + (j == query_b[2]) + (k == query_b[2]) + (l == query_b[2]) + (m == query_b[2]) + (n == query_b[2])
    count_b3 = (i == query_b[3]) + (j == query_b[3]) + (k == query_b[3]) + (l == query_b[3]) + (m == query_b[3]) + (n == query_b[3])
    if count_b1 >= 1 && count_b2 >= 1 && count_b3 >= 1
        count_bb += 1
    end
end

println("Total number of strings generated: $(10^6)")
println("\n------------------------------")
println("Count of strings containing $(query_a)$(query_a)$(query_a): $count_aa = $(count_aa / 10^6 * 100)%")
println("Count of strings containing $(query_b[1])$(query_b[2])$(query_b[3]): $count_ba = $(count_ba / 10^6 * 100)%")
println("\n------------------------------")
println("Count of strings containing three $query_a's anywhere: $count_ab = $(count_ab / 10^6 * 100)%")
println("Count of strings containing $(query_b[1]), $(query_b[2]), and $(query_b[3]) in any order: $count_bb = $(count_bb / 10^6 * 100)%")
```
